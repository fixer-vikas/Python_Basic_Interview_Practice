```python
num = int(input("Enter number: "))
rev = 0
while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10
print("Reversed number:", rev)
```

```python
num = int(input("Enter number: "))
count = 0
while num > 0:
    num = num // 10
    count += 1
print("Digits:", count)
```

```python
num = int(input("Enter number: "))
sum = 0
while num > 0:
    digit = num % 10
    sum += digit
    num //= 10
print("Sum of digits:", sum)
```

```python
num = int(input("Enter number: "))
temp = num
rev = 0
while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num //= 10
if temp == rev:
    print("Palindrome")
else:
    print("Not palindrome")
```

```python
n = int(input("Enter terms: "))
a = 0
b = 1
for i in range(n):
    print(a)
    c = a + b
    a = b
    b = c
```

```python
num = int(input("Enter number: "))
prime = True
for i in range(2, num):
    if num % i == 0:
        prime = False
        break
if prime:
    print("Prime")
else:
    print("Not prime")
```

```python
for num in range(2, 51):
    prime = True
    for i in range(2, num):
        if num % i == 0:
            prime = False
            break
    if prime:
        print(num)
```

```python
numbers = [3, 7, 2, 9, 4]
largest = numbers[0]
for n in numbers:
    if n > largest:
        largest = n
print("Largest:", largest)
```

```python
numbers = [3, 7, 2, 9, 4]
smallest = numbers[0]
for n in numbers:
    if n < smallest:
        smallest = n
print("Smallest:", smallest)
```
