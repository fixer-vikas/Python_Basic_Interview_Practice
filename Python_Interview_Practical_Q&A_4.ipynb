# 3. Count even numbers in list

```python
numbers = [1,2,3,4,5,6]
count = 0
for n in numbers:
    if n % 2 == 0:
        count += 1
print("Even numbers:", count)
```

# 4. Count odd numbers in list

```python
numbers = [1,2,3,4,5,6]
count = 0
for n in numbers:
    if n % 2 != 0:
        count += 1
print("Odd numbers:", count)
```

# 9. Check palindrome string

```python
text = input("Enter string: ")
print('This is the input string : ', text)
if text == text[::-1]:
    print("Palindrome")
else:
    print("Not palindrome")
```

# 10. Count vowels in string

```python
text = input("Enter text: ")
print('This is the input text : ', text)
count = 0
for ch in text:
    if ch.lower() in "aeiou":
        count += 1
print("Vowels:", count)
```

# 16. Check Armstrong number

```python
num = int(input("Enter number: "))
print('This is the input number : ', num)
sum = 0
temp = num
while temp > 0:
    digit = temp % 10
    sum += digit ** 3
    temp //= 10
if sum == num:
    print("Armstrong")
else:
    print("Not Armstrong")
```

# 20. Check leap year

```python
year = int(input("Enter year: "))
print('This is the input year : ', year)
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print("Leap year")
else:
    print("Not leap year")

```
